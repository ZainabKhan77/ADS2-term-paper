```r
library(dplyr)
library(plm)
data("Grunfeld" , package = "plm")

#select specific columns
Grunfeld_selected<- Grunfeld %>% select(firm,value,capital)
print(head(Grunfeld_selected))

#filter rows based on a condition
Grunfeld_filtered<- Grunfeld %>% filter(year==1945)
print(head(Grunfeld_filtered))

#create a new column (mutate)
Grunfeld_mutate<- Grunfeld %>% mutate(inv_per_year=inv/year)
print(head(Grunfeld_mutate))

#arrange rows by a column
Grunfeld_arranged<- Grunfeld %>% arrange(capital,firm)
print(head(Grunfeld_arranged))

#summarize data (eg calc mean capital)
Grunfeld_summarized<- Grunfeld %>% summarise(mean_capital= mean(capital))
print(head(Grunfeld_summarized))

#mean capital per year
Grunfeld_by.year<- Grunfeld %>% group_by(year) %>% summarise(mean_capital=mean(capital))
print(head(Grunfeld_by.year))

#transforming variables
df <- Grunfeld # Assign Grunfeld to df first
df <- df %>% mutate(inv= log(inv), value= log(value), capital= log(capital))
# The line below was causing the error due to incorrect syntax.
# df <- df %>% mutate(inv= log(inv)* value= log(value)* capital= log(capital))
print(head(df))

#convert to panel data format
pdata <- pdata.frame(Grunfeld, index = c("firm", "year"), drop.index = FALSE)

#fixed effects (FE) model - within estimator
fe_model <- plm(inv ~ value + capital, data = pdata, model = "within")
summary(fe_model)
```
